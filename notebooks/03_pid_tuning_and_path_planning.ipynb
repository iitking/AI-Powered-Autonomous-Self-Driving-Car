## 1. PID Formulation
$$u(t) = K_p \cdot e(t) + K_i \int_0^t e(\tau)d\tau + K_d \frac{de(t)}{dt}$$
- $K_p$ (Proportional): Reacts proportionally to current cross-track error
- $K_i$ (Integral): Eliminates persistent steady-state bias (e.g. road crown or steering misalignment)
- $K_d$ (Derivative): Dampens oscillations and overshooting during rapid corrections

```python
import sys
sys.path.append('..')

from src.control.pid_controller import PIDController
from src.navigation.path_planner import GridAStar

pid = PIDController(kp=0.45, ki=0.01, kd=0.12, output_limits=(-30.0, 30.0))
print("Simulating step response for 10 cm lateral lane offset:")

cte = 0.10  # 10 cm error
for step in range(5):
    steer_deg = pid.update(target=0.0, current=cte, dt=0.05)
    print(f"Step {step+1}: Steer Output = {steer_deg:.2f}° | P={pid.last_telemetry.p_term:.2f}, I={pid.last_telemetry.i_term:.2f}, D={pid.last_telemetry.d_term:.2f}")
    cte *= 0.7  # Error decays as car steers back to center
```

## 2. A* Path Planner Demonstration
Planning an optimal collision-free trajectory around obstacles on a 5m x 5m grid:

```python
planner = GridAStar(width_m=5.0, height_m=5.0, resolution_m=0.1)
# Place obstacle right in the center
planner.set_obstacle(x=2.5, y=2.5, radius_m=0.3)

start = (0.5, 2.5)
goal = (4.5, 2.5)
path = planner.plan(start, goal)

print(f"Generated collision-free path with {len(path)} waypoints:")
for pt in path:
    print(f"  Waypoint: ({pt[0]:.2f}, {pt[1]:.2f})")
```
