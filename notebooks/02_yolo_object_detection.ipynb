## 1. Monocular Camera Distance Estimation
Given an object of known physical height $H_{\text{real}}$, focal length $f$, and bounding box pixel height $h_{\text{pixel}}$:
$$d = \frac{f \cdot H_{\text{real}}}{h_{\text{pixel}}}$$
Lateral offset $y_{\text{lat}}$ from optical axis:
$$y_{\text{lat}} = \frac{(c_x - \text{center}_x) \cdot d}{f}$$
